Customer Churn Data Analysis & Visualization
I recently worked on a dataset of more than 10,000 credit card customers, with the objective of analyzing customer behavior and identifying the key factors influencing retention and attrition.

📊 The dataset included:

Financial information (Income, Credit Limit, Average Spend).

Activity indicators (Transaction Count, Transaction Amount, Utilization Ratio).

🛠️ Approach:

Conducted a comprehensive Exploratory Data Analysis (EDA) to uncover patterns and insights.

Applied Data Visualization techniques using Python (Seaborn, Matplotlib) to highlight relationships across variables.

📈 Key Findings:

Customers with higher transaction activity and utilization levels were more likely to remain with the bank.

Low-tier cardholders (e.g., Blue) and customers with limited usage showed higher attrition rates.

Demographics such as age and education level demonstrated a notable impact on customer behavior.

💡 Conclusion:
This analysis provided valuable insights that can support financial institutions in developing effective customer retention strategies and enhancing loyalty.

***Data visualization***

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

**Age Distribution**

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Customer_Age"],bins=20,kde=True)
plt.title("Age Distribution")
plt.show()
```

**Distribution (Attrition_Flag)**

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df,x="Attrition_Flag")
plt.title(" Distribution (Attrition_Flag)")
plt.show()
```

**Education_Level vs Attrition_Flag**

```python
plt.figure(figsize=(8,5))
des1=df["Education_Level"].value_counts().index
sns.countplot(data=df,x="Education_Level",hue="Attrition_Flag",order=des1)
plt.xticks(rotation=45)
plt.title("Education_Level vs Attrition_Flag")
plt.show()
```

**Marital_Status Distribution**

```python
plt.figure(figsize=(6,6))
Marital_Status_counts = df["Marital_Status"].value_counts()
plt.pie(
   Marital_Status_counts,
    labels=Marital_Status_counts.index,
    autopct="%1.1f%%",
    startangle=140,
    colors=sns.color_palette("Set2", len(Marital_Status_counts))
)
plt.title("Marital_Status Distribution")
plt.show()
```

**Avg_Utilization_Ratio By Gender**

```python
sns.boxplot(data=df, x="Gender", y="Avg_Utilization_Ratio", palette="Set2", showfliers=False)
plt.title("Avg_Utilization_Ratio By Gender")
plt.show()
```
